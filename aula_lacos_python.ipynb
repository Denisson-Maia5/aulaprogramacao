```python
#Mostras a soma todos os numeros de um intervalo, inserido pelo usuario
soma = 0
for i in range (5):
  soma = soma = i
print(soma)
```

```python
#Mostras a soma todos os numeros de um intervalo, inserido pelo usuario
soma = 0
num1 = int(input("insira o primeiro numero: "))
num2 = int(input("insira o final do intervalo: "))
for i in range (num1, num2+1):
  soma = soma + i
print(soma)
```

```python
#Pega um range de valore entre 1 e 20.
#E imprime informando se é par ou impar
#Ex: 1 é impar
for i in range (1, 21):
  if (i % 2 == 0):
   print(i, "é par")
  else:
   print(i, "é impar")
```
