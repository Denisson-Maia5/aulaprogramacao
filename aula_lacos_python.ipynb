```python
#Escreva um game para encontrar o número secreto gerado.
#O sistema gera um número aleatorio entre 0 e 100
#O usuario dar um palpite
#quando o palpite for errado:
##### Se for maior que o número, esse se torna o número maximo
##### Se for menor que o número, esse se torna o número minimo
#O jogo termina quando a pessoa acertar o numero.
## BONUS: Mostre em quantas tentativas conseguiu acertar.
numero_secreto = random.randint(0, 100)
print(numero_secreto)
minimo = 0
maximo = 100
print(f"o numero secreto está entre {minimo} e {maximo}")
palpite = int(input(""))
while(palpite != numero_secreto):
  print("errou")
  if(palpite > numero_secreto):
    maximo = palpite
  else:
    minimo = palpite
  print(f"digite um numero entre {minimo} e {maximo}")
  palpite = int(input(""))

print("acertou")



```
