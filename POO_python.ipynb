```python
class Aluno:
  def __init__(self, nome, ra, curso):
   self.nome = nome
   self.ra = ra
   self.curso = curso
  def apresentar(self):
    print(f'Sou {self.nome}, {self.ra}, {self.curso}')
```

```python
aluno1 = Aluno('Joaquim', '123456456', 'ADS')

print(aluno1.nome)
print(aluno1.ra)
print(aluno1.curso)
aluno1.apresentar()
```

```python
aluno2 = Aluno('Joaquim', '123456456', 'ADS')

print(aluno1)
print(aluno2)

print(aluno1 == aluno2)

```

```python
#Crie uma classe Animal com os atributos:
#nome, especie
#método emitir_som()

class Animal:
  def __init__(self, nome, especie, som):
    self.nome = nome
    self.especie = especie
    self.som = som
  def emitir_som(self):
    print(self.som)

```

```python
gato = Animal('trombadinha', 'vira lata', 'Miau')
gato.emitir_som()

```

```python
#Crie uma classe carro com os atributos
#marca, modelo e ano. e um método descrição()
#que imprime uma frase apresentando o carro.
#insira outro metodo chamado freiar()
#qundo for chamado imprime "o carro freiou"

class Carro:
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano

  def descricao(self):
     print(f'marca:{self.marca}, modelo:{self.modelo}, Ano:{self.ano}')

  def freiar(self):
    print('O carro freiou')
```

```python
carro1 = Carro('Opala', 'Chevrolet', '76')
carro1.descricao()
carro1.freiar()
```

```python
class Animal:
  def __init__(self, nome):
    self.nome = nome
  def emitirsom(self):
    print('som generico de animal')
```

```python
class Cachorro(Animal):
  def emitirsom(self):
    print('AU! AU!')

class Gato(Animal):
  def emitirsom(self):
    print('Miau Miau')
```

```python
dog = Cachorro('Bob')
cat = Gato('trombadinha')
dog.emitirsom()
cat.emitirsom()
```

```python
#Criar uma classe funcionario
#Criar subclasses Gerente e Vendendor
#A classe funcionario tem atributos
#nome e salario
#metodo calculaSalario()
#A classe gerente tem o nome do setor

class Funcionario:
  def __init__(self, nome, salario):
    self.nome = nome
    self.salario = salario

  def calculaSalario(self):
    print(self.salario)

```

```python
class Gerente(Funcionario):
  def __init__(self, setor):
    self.setor = setor
  def calculaSalario(self):
    print(self.salario * 1.5)
```

```python
gerente = Gerente('logistica')
gerente.salario = 5000
gerente.calculaSalario()
```
